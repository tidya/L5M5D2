```python
import pandas as pd


# Load the data.
raw = pd.read_csv(
    r"C:\Users\Admin\Desktop\L5M5D2\Data\Library Systembook.csv"
)


class Cleaner:
    def __init__(self, df):
        self.df = df.copy()

    def CleanCheckoutDates(self):
        self.df["Checkout_Cleaned"] = (
            self.df["Book checkout"]
            .str.replace('"', "", regex=False)
            .str.replace("\\", "", regex=False)
            .str.strip()
        )

    def CleanBadDates(self):
        self.df["Checkout_Cleaned"] = pd.to_datetime(
            self.df["Checkout_Cleaned"],
            format="%d/%m/%Y",
            errors="coerce"
        )

        self.df["Book_Cleaned"] = pd.to_datetime(
            self.df["Book Returned"],
            format="%d/%m/%Y",
            errors="coerce"
        )

    def CleanBookTitles(self):
        self.df["Books"] = self.df["Books"].str.strip()

    def CleanMissingValues(self):
        self.df = self.df.dropna(how="all")
        self.df = self.df.dropna(subset=["Id"])

    def CleanDuplicates(self):
        loan_columns = [
            "Books",
            "Checkout_Cleaned",
            "Book_Cleaned",
            "Days allowed to borrow",
            "Customer ID"
        ]
        self.df = self.df.drop_duplicates(subset=loan_columns)

    def ShowBadDates(self):
        bad_dates = self.df["Checkout_Cleaned"] > self.df["Book_Cleaned"]

        print("\nCheckout after return:")
        print(
            self.df.loc[
                bad_dates,
                ["Id", "Books", "Checkout_Cleaned", "Book_Cleaned"]
            ].to_string(index=False)
        )

    def ShowInvalidDates(self):
        invalid_dates = (
            self.df["Checkout_Cleaned"].isna()
            | self.df["Book_Cleaned"].isna()
        )

        print("\nMissing or invalid dates:")
        print(
            self.df.loc[
                invalid_dates,
                ["Id", "Books", "Book checkout", "Book Returned"]
            ].to_string(index=False)
        )

    def ShowMissingValues(self):
        missing_values = (
            self.df["Books"].isna()
            | self.df["Customer ID"].isna()
        )

        print("\nMissing book title or customer ID:")
        print(
            self.df.loc[
                missing_values,
                ["Id", "Books", "Customer ID"]
            ].to_string(index=False)
        )

    def ShowColumns(self):
        print(self.df.columns)

cleaner = Cleaner(raw)

cleaner.CleanMissingValues()
cleaner.CleanCheckoutDates()
cleaner.CleanBadDates()
cleaner.CleanBookTitles()
cleaner.CleanDuplicates()

cleaner.ShowBadDates()
cleaner.ShowInvalidDates()
cleaner.ShowMissingValues()

print("\nCleaned data:")
print(cleaner.df.to_string(index=False))
```
